```python
class pet:
    def __init__(self, name,age):
        self.name = name
        self.age =age
       

    def message(self):
        print(f"My name is {self.name}")
        print(f"My age is {self.age}")
    
    def sleep(self):
        print("sleeping")

    def wakeup(self):
        self.sleep()
        print("wakeup")    

class dog(pet):
    def __init__(self,name,age,breed):
        super().__init__(name,age)
        self.breed=breed

    def speak(self):
        print(f"I am a dog of breed {self.breed}")
        print("I bark") 

class cat(pet):

    def __init__(self,name,age,breed):
        super().__init__(name,age)
        self.breed=breed

    def speak(self):
        print("I am a Cat")
        print("I meaow") 
```

```python
d2=dog("ceaser",2,"husky")
d2.message()
d2.speak()
d2.wakeup()
```
