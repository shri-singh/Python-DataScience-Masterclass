```python
## Why Funcions are required
def even_odd(num):
    if num%2==0:
        return "Even"
    else:
        return "Odd"
    
print(even_odd(5))
print(even_odd(6))
```

```python
def div_test(dividend,divisor):
    if dividend % divisor ==0:
        return 0
    else:
        return 1
    
prime_numbers = [ x for x in range(100) for j in range(2,x) if div_test(x,j)==1]    
```
