```python
def prime_test(dividend):
    if dividend >1:    
        for test_all_nums in range(2,dividend):
            if dividend % test_all_nums ==0:
                return 0
                break
        else:
            return 1
    else:
        return 0
prime_numbers = [ x for x in range(100) if prime_test(x)==1]    
```

```python
prime_numbers
```

```python
print(f"Prime test for number {10} is {prime_test(10)} ")
```

```python
def prime_test_print(x):
    if prime_test(x):
        print(f" Number {x} is a Prime number")
    else:
        print(f"number {x} is not a Prime number")

prime_test_print(29)
prime_test_print(30)
```

# OOPS
## Paradiagm for abstraction of properties and methods -> Class
## Instnces of class - > Object
## Object.property
### Self referencing during definitiation phase for 



```python
class shape():
    #len , width =1,10
    #def _
    def __init__(self,len,width):
        self.len=len
        self.width = width

    def area(self):
        return self.len* self.width

    def paint_cost(self, cost_perunit):
        return self.area()*cost_perunit
```

```python
s = shape(3,5)
print (f"Area of shape is {s.area()}")
print (f"Cost of painiting this shape is {s.paint_cost(3)}")
```
