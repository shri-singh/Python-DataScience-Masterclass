# Control Flow :
### decides which lines of code to execute , in what order and how many times based on conditions and loops and functions calls.

## 3 Main types of control flow :

### 1. Sequential : the default flow of execution where code is executed line by line from top to bottom.
### 2. Conditional - Different blocks of code are executed based on certain conditions being true or false.
####   (if, elif, else)
### 3. Iterative: Repeat code blocks multiple times based on a condition or a sequence of items.
####    Loops (for [over time], while[until condition is True])
####    Iterators (iter, next)
####    Loop control statements (break [exit loop early], continue[skip to next iteration], pass[do nothing/placeholder for future code])

## Functions: reusable blocks of code that perform a specific task and can be called multiple times with different inputs.
####    Change control flow by jumping into block of code when function is called and returning back to the point of call after execution.
####    Function definition (def), function call, return statement, parameters and arguments, scope and lifetime of variables within functions.

## Exceptions: Error bsed control flow that allows programs to handle and respond to errors gracefully without crashing. 
####   try-except blocks, raising exceptions, handling specific exception types, finally block for cleanup actions, custom exception classes.

# In Short : Control flow = decisions + iterations + early exits + function calls + exception handling
