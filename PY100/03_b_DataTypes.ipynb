# DataTypes:
## Data types matter because they decide what operations are allowed, how values behave in memory, and how you model data cleanly 
#### ( helps avoid bugs like wrong comparisons, mutability surprises, or type errors).

## 1. Numeric
#### int:   whole numbers, unlimited precision
#### float: decimal (IEEE-754), rounding issues possible
#### complex: a + bj
#### bool: True/False (subclass of int: True==1, False==0)


```python
x= 100 #this is easy integer assignment 
## for floating point numbers, max precision is 15 digits
y= 3.1415926535897932384626433832795 #this is a floating point number assignment
z= 1.0e-10 #this is a floating point number in scientific notation
print("x =", x)
print("y =", y)
print("z =", z)
```

## Floating point bit precision is always 64 bit iEEE-754 double pression
#### in 64 bit layout : 1 bit is sign, 11 bits is for exponents and 52 bits is fraction(mantissa)
#### Q- why 53 bits translate to 16 digits => because in decimal we are using base 10
#### so its 53.log2(base10)


```python

import math
print(53*math.log10(2)) #this is a mathematical expression using the logarithm function 
n = 2**53
print(float(n) == float(n + 1))  # True  ✅ they become indistinguishable
print(float(n), float(n + 1))    # prints the same value
print(float(n + 2) == float(n))  # False (step size here is 2)
```

### For curious ones only-
#### Complex Number operations


```python
z1= 1 + 2j #this is a complex number assignment
z2= complex(1, 2) #this is another way to assign a complex number
print("z1 =", z1)
print("z2 =", z2)   

z3 = z1 + z2 #this is a complex number addition
print("z3 =", z3)

z4 = z1 * z2 #this is a complex number multiplication
print("z4 =", z4)
```
