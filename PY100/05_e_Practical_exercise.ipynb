```python
# Use a dictionary to count the frequency of elements in list
numbers =[1,2,2,3,2,2,3,3,3,4,4,4,2,3,2,1,3,4]
set_unique =set(numbers)
dict_out ={}
for i in set_unique:
    dict_out[i]=numbers.count(i)
print (dict_out)
```

```python
# Using first  principle
freq={}
for i in numbers:
    if i in freq:
        freq[i] +=1
    else:
        freq[i]=1

print(freq)        
```

```python
freq2 ={x:numbers.count(x) for x in set_unique}
freq2
```
