# Operators : Perform action on memory location or variables

## Airthemetic Operators
###    Used for Mathematical operations


```python
a, b = 10, 3
a + b   # 13  addition
a - b   # 7   subtraction
a * b   # 30  multiplication
a / b   # 3.333... true division (always float)
a // b  # 3   floor division
a % b   # 1   modulus (remainder)
a ** b  # 1000 exponent (power)
```

## AirtAssignment Operators
###   assign value to variable

```python
x = 5
x += 2   # x = x + 2  -> 7
x -= 1   # -> 6
x *= 3   # -> 18
x /= 2   # -> 9.0
x //= 2  # -> 4.0 (careful: type may change earlier)
x %= 3   # -> 1.0
x **= 2  # -> 1.0
```

```python
a, b = 1, 2      # multiple assignment
a, b = b, a      # swap
```

## Operator Precedence
    #### ()
    #### **
    #### * / // %
    #### + -
    #### comparisons (==, <, >, …)
    #### not
    #### and
    #### or
### When in doubt: use parentheses for clarity.

```python
## Operator Precedence
result = 3 + 4 * 2 / (1 - 5) ** 2 ** 3
print(result)  # 3.0001220703125

```

# Comparision Operators
## compare two expressions and return a boolean value


```python
number1 = 45
number2 = 45
print(number1 == number2)  # True (values are equal)
print(number1 is number2)  # True (small integers are cached by Python) 
```

```python
number1 = 45
number2 = 55

print(number1 <= number2)  # False (values are not equal)
print(number1 >= number2)  # False (values are not equal)
print(number1 < number2)   # True (45 is less than 55)
```

```python
# Logical Operators
a = True    
b = False
print(a and b)  # False (both must be True)
print(a or b)   # True (at least one is True)

b=True
print(a and b)  # True (both are now True)
```
