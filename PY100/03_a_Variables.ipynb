## Type Checking and Conversion
### Type conversion / type casting in Python = changing a value from one data type to another.

## 1. Implicit Conversion
### (automatic): Python converts only when it’s safe (mostly numeric “widening”).
#### int → float, int/float → complex in mixed math.
#### Example: 3 + 2.5 → 5.5 (int becomes float)

## 1. Explicit Conversion
### Manual/Forces: you use functions like int(), float(), str(), list(), etc.
#### You can only force compatible (convertible) data types

```python
## Type Casting Rules
x = 10.50
print(f" Type of x is {type(x)} and value is {x}")
y = int(x)
print(f" Type of y is {type(y)} and value is {y}")  

w =float(y)
print(f" Type of w is {type(w)} and value is {w}")

z=str(x)
print(f" Type of z is {type(z)} and value is {z}")

x="I am going to do trick you"
print(f" Type of x is {type(x)} and value is {x}")
## y = int(x)  # This will raise an error because the string cannot be converted to an integer

print(f" Type of y is {type(y)} and value is {y}")  
z=str(x)
```
