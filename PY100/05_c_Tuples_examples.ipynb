```python
# list can be converted into tuples and tuples can be converted back to list

list=[1,2,3,4,5]
print(type(list))
tuple_eg = tuple(list)
print(tuple_eg)
print(type(tuple_eg))
```

```python
print(tuple_eg[::])
# similar slicing operations as list, just return type is tuple
```

```python
## Tuple operations
#concatenation using + operation
mixed_t = ("hello", "world")
concat_tuple = mixed_t + tuple_eg
concat_tuple
```

```python
mult_tuple = mixed_t*3
mult_tuple
```

```python
# immutable nature of tuples
tuple_eg[2]=10 # will give an eror
```

## Methods of Tuples
1. tuple.count(element) -- Counts the number of times element is found in tuple
2. tuple.index(element ) -- index of first occurance of element


```python
## Packing and uppacking of tuples

numbers = 3,4.5,"Hello",10,4,5,5,10
first, *middle,last =numbers
print(first)
print(middle)
print(last)
```
