# Sets 
## Sets are in-built data type in Python - to store collections of unique items
## Unordered (items do not follow any arrangement of ascending or descending)
## Do not allows duplicate elements
##  used for membership tests, eliminating duplicate values, performing set operations like union, intersection difference and symmetric difference



```python
# creating sets using {}, just like dictionary its not key value pair
set1 ={1,2,3,4,5,8}
print(set1)
print(type(set1))
```

```python
# creating empty sets
set2={} # this will create empty dictionary-- be carefull
set3 =set()

print(set2,set3)
print(type(set2),type(set3))
```

```python
#Basic set operations

# Adding new elements using add method
set3.add(3)
set3.add(5)
set3.add(7)
set3.add(8)
set3.add(9)
set3.add(8) # will not add as its already there
set3
```

```python
## Removing element from set

set3.discard(9)
set3
```

```python
set3.pop() # pop is first in first out fifO
set3
```

```python
# Membership test - using in operator
print(10 in set3)
print(5 in set3)
```

```python
# set operations - Union, intersection, difference, symmetric difference, subset, superset
set1 ={1,2,3,4,5,6,7}
set2 ={4,5,6,7,8,8,9,10}

union_set =set1.union(set2)  #combines all the unique elements form a and b set
print(union_set)
intersect_set =set1.intersection(set2) # takes the common elemets from a and b
print(intersect_set)

difference_set =set1.difference(set2) #keep a and remove common elements form b
print(difference_set)

difference_set =set2.difference(set1) #keep a -b is not same as b-a
print(difference_set)

result =set1.symmetric_difference(set2) # keep union - the intersection results.
print(result)

result =set2.symmetric_difference(set1)
print(result)
```

```python
set1.intersection_update(set2) # updates in place set operaitons
set1
```

```python
print(set1.issubset(set2))
```

```python
print(set2.issuperset(set1))
```
